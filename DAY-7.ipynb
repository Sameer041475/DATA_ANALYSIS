```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("ball_by_ball_ipl.csv")
df
```

```python
df['Match ID']
```

```python
df['year'] = pd.to_datetime(df['Date']).dt.year
```

```python
df
```

```python
final_matches = df.groupby('year')['Match ID'].max()
```

```python
df[df['Match ID'].isin(final_matches)][['Match ID', 'year','Winner']].drop_duplicates()['Winner'].value_counts()
```

```python
winner_counts = df[df['Match ID'].isin(final_matches)][['Match ID', 'year','Winner']].drop_duplicates()['Winner'].value_counts().reset_index()
winner_counts
winner_counts.columns = ['Team', 'Titles Won']
```

```python
winner_counts.columns
```

```python
plt.figure(figsize = (10,6))
sns.barplot(x = 'Team',y = 'Titles Won', data = winner_counts,palette = 'viridis')
plt.title('Number of IPL titles won by each team',fontsize = 16)
plt.xlabel("Team", fontsize = 12)
plt.ylabel("Title Won", fontsize = 12)
plt.xticks(rotation = 45)
plt.show
```

```python
chase_stats = (
    df[['Match ID', 'year', 'Chased Successfully']]
    .drop_duplicates()
    .groupby(['year', 'Chased Successfully'])
    .size()
    .reset_index(name='Matches')
)

chase_stats
```
