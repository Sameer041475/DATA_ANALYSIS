## Types of indexing

```python
df['Batter']
```

```python
df[['Batter','Winner']]
```

# .loc[] - Label based Selection

```python
df.loc[[239688,239692]]
```

```python
df.loc[:,'Over']
```

```python
df.loc[:,['Bat First','Venue']]
```

```python
df.loc[:,'Match ID':'Batter']
```

```python
# .loc[]- Conditional Row Selection
df.loc[df['Method'] == 'lbw']
```

# .iloc[]- Position based indexing

```python
# .iloc[] -- single row selection
df.iloc[2]
```

```python
df.iloc[[2,3,4,5]] # multi label selection 
```

```python
# .iloc -- column selection
df.iloc[:,2:5]
```

```python
# row slicing
df.iloc[1:3]
```

```python
# .iloc[]-- Row and column selection
df.iloc[1:3,0:3]
```

```python

.iloc[] VS .iloc[]

Feature

Indexing Type

Slicing Behavior

Conditional Selection

Use Cases

                    .Joc[] (Label-based)
                    
                    Uses LABELS for indexing
                    
                    Inclusive of both start and end position
                    
                    Can use boolean conditions directly
                    
                    Use when selecting by label

                                                            .iloc[] (Position-based)
                                                            
                                                            Uses INTEGER POSITIONS
                                                            
                                                            Excludes the end position
                                                            
                                                            Cannot use boolean conditions
                                                            
                                                            Use when selecting by position
```
